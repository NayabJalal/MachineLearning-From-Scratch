```python
import numpy as np

from sklearn.linear_model import SGDRegressor
from sklearn.metrics import mean_absolute_error
```

```python
X = np.array([
    [500],
    [1000],
    [1500],
    [2000],
    [2500],
    [3000],
])

y = np.array([
    [15],
    [20],
    [25],
    [30],
    [35],
    [40],
])
```

```python
batch_model = SGDRegressor(
    max_iter=2000,
    tol=1e-6,
    random_state=42
)
batch_model.fit(X , y)

prediction = batch_model.predict([[1800]])

print("Batch Prediction" , prediction[0])
```

```python
online_model = SGDRegressor(
    learning_rate="constant",
    eta0=0.000001,
    random_state=42
)


for i in range(len(X)):
    online_model.partial_fit(
        X[i:i+1],
        y[i:i+1]
    )

# Predict
prediction = online_model.predict([[1800]])

print("Online Prediction:", prediction[0])
```
