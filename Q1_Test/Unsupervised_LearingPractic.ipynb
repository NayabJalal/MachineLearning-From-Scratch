```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.cluster import KMeans
```

```python
data = {
    "Annual_Income": [
        15, 18, 20, 22, 25,
        60, 65, 70, 75, 80,
        120, 125, 130, 135, 140
    ],

    "Spending_Score": [
        80, 85, 78, 82, 88,
        40, 45, 42, 48, 44,
        20, 25, 18, 22, 28
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python
plt.scatter(
    df["Annual_Income"],
    df["Spending_Score"]
)

plt.xlabel("Annual Income")
plt.ylabel("Spending Score")

plt.title("Customer Data Before Clustering")

plt.show()
```

```python
# Select features---
X = df[["Annual_Income", "Spending_Score"]]

model = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

model.fit(X)

df["Cluster"] = model.labels_

print(df)
```

```python
plt.scatter(
    X["Annual_Income"],
    X["Spending_Score"],
    c=df["Cluster"],
    s=100
)

plt.xlabel("Annual Income")
plt.ylabel("Spending Score")

plt.title("Customer Segmentation Using K-Means")

plt.show()
```
