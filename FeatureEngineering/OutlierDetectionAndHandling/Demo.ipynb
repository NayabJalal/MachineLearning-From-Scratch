```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df= pd.DataFrame({
    "Salary" : [
        20000,22000,25000,27000,
        30000,32000,35000,38000,
        40000,500000
    ]
})
```

```python
sns.boxplot(x=df["Salary"])
plt.title("Before Outlier Handling")
plt.show()
```

```python
#DETECT OUTLIERS ->
Q1 = df["Salary"].quantile(0.25)
Q3 = df["Salary"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[
    (df["Salary"] < lower) | 
    (df["Salary"] > upper)
]

print("Outliers : ")
print(outliers)
```

```python
#Applying capping->
df["Salary_Capped"] = df["Salary"].clip(
    lower=lower,
    upper=upper
)

print("\nAfter Capping : ")
print(df)
```

```python
sns.boxplot(x=df["Salary_Capped"])
plt.title("After Outlier Handling")
plt.show()
```
