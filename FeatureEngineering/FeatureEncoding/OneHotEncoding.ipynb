OneHotEncoding with Top Categories --->>>

```python
counts = df['brand'].value_counts()
```

```python
df['brand'].nunique()
threshold = 100
```

```python
replace = counts[counts <= threshold].index
```

```python
pd.get_dummies(df['brand'].replace(replace , 'others'), dtype=int)
```

Problem here is we are explicitly Breaking into two parts and merging using HSTACK, to solve this problem Column Transformer Comes into picture :>

```python
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
```

```python
df.head()
```

```python
transformer = ColumnTransformer(transformers=[
    ('tnf1' ,SimpleImputer(), ['km_driven']),
    ('tnf2' ,OrdinalEncoder(categories=[['First Owner' , 'Second Owner' , 'Third Owner' , 'Fourth & Above Owner', 'Test Drive Car']]),['owner']),
    ('tnf3' ,OneHotEncoder(sparse_output=False,drop='first'),['brand' , 'fuel'])
],remainder='passthrough') #remainder='drop' for dropping the first col.
```

```python
# transformer.fit_transform(X_train) -> In numpy, For DF 
transformer.set_output(transform="pandas")

X_train_transformed = transformer.fit_transform(X_train)

X_train_transformed
```
