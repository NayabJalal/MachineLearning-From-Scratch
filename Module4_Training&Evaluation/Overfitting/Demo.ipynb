```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
```

```python
data = load_breast_cancer()
X, y = data.data, data.target
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)
```

```python
underfit_model = DecisionTreeClassifier(max_depth=1, random_state=42)
underfit_model.fit(X_train, y_train)
```

```python
good_model = DecisionTreeClassifier(max_depth=4, random_state=42)
good_model.fit(X_train, y_train)
```

```python
print("--- Underfitted Model (max_depth=1) ---")
print(f"Train Accuracy: {underfit_model.score(X_train, y_train):.3f}")
print(f"Test Accuracy : {underfit_model.score(X_test, y_test):.3f}")

print("\n--- Good Fit Model (max_depth=4) ---")
print(f"Train Accuracy: {good_model.score(X_train, y_train):.3f}")
print(f"Test Accuracy : {good_model.score(X_test, y_test):.3f}")
```

```python
labels = ['Underfit (Depth=1)', 'Good Fit (Depth=4)']
train_scores = [
    underfit_model.score(X_train, y_train),
    good_model.score(X_train, y_train)
]
test_scores = [
    underfit_model.score(X_test, y_test),
    good_model.score(X_test, y_test)
]
```

```python
x = [0, 1]
width = 0.35

plt.figure(figsize=(8, 5))
plt.bar([i - width/2 for i in x], train_scores, width=width, label='Train Accuracy', color='steelblue')
plt.bar([i + width/2 for i in x], test_scores, width=width, label='Test Accuracy', color='coral')

plt.ylim(0.80, 1.02)
plt.xticks(x, labels, fontsize=11)
plt.ylabel('Accuracy Score', fontsize=12)
plt.title('Underfitting vs. Good Fit on Breast Cancer Dataset', fontsize=13)
plt.legend()
plt.grid(axis='y', alpha=0.3)
plt.show()
```
