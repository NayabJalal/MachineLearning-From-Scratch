```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "StudyHours": [2, 4, 6, 8],
    "SleepHours": [8, 7, 6, 5],
    "Marks": [40, 55, 75, 90]
})
```

```python
df["TotalProductiveHours"] = (
    df["StudyHours"] + (8 - df["SleepHours"])
)
```

```python
df["MarksPerStudyHour"] = (
    df["Marks"] / df["StudyHours"]
)
```

```python
print(df)
```
