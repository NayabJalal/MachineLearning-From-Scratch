```python
import pandas as pd
from sklearn.impute import SimpleImputer
```

```python
df = pd.read_csv("students.csv")
```

```python
print(df.isnull().sum())
```

```python
num_imputer = SimpleImputer(strategy="median")
```

```python
df[["SleepHours", "StudyHours", "Attendance", "Marks"]] = num_imputer.fit_transform(
    df[["SleepHours", "StudyHours" , "Attendance", "Marks"]]
)
```

```python
cat_imputer = SimpleImputer(strategy="most_frequent")
```

```python
df[["City"]] = cat_imputer.fit_transform(df[["City"]])
```

```python
print(df.isnull().sum())
```
